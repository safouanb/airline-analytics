Logistic Regression Model (all features)

```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score

from preprocessing import load_data, get_preprocessor


#load train data

X_train, y_train = load_data("train.csv", data_dir='data/raw')

preprocessor = get_preprocessor(X_train)

#build Logistic Regression pipeline

log_reg_full = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("scaler", StandardScaler(with_mean=False)),  # needed for LR after one-hot encoding
    ("model", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        solver="liblinear"
    ))
])

#cross-validation (5-fold)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

metrics = ["accuracy", "f1_macro", "roc_auc"]

print("Logistic Regression with all features \n")

for metric in metrics:
    scores = cross_val_score(log_reg_full, X_train, y_train, cv=cv, scoring=metric)
    print(f"{metric} scores:", scores)
    print(f"Mean {metric}: {scores.mean():.4f} (std: {scores.std():.4f})\n")
```

Logistic Regression with Feature Selection (7 Features)

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

#select features
selected_features = [
    'Online boarding',
    'Inflight wifi service',
    'Type of Travel',
    'Class',
    'Inflight entertainment',
    'Customer Type',
    'Leg room service',
    'satisfaction' #target
]


#load train.csv
df = pd.read_csv("train.csv")
df = df[selected_features]

#separate target 
#y = 1 if satisfied, 0 if neutral/dissatisfied

X = df.drop("satisfaction", axis=1)
y = df["satisfaction"].apply(lambda x: 1 if x == "satisfied" else 0)

X = pd.get_dummies(X, drop_first=True)

log_reg = Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",   #handles slight imbalance
        solver="liblinear"
    ))
])


#cross-validation (5-fold)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

metrics = ["accuracy", "f1_macro", "roc_auc"]

print("Logistic Regression with 7 features \n")

for metric in metrics:
    scores = cross_val_score(log_reg, X, y, cv=cv, scoring=metric)
    print(f"{metric} scores:", scores)
    print(f"Mean {metric}: {scores.mean():.4f}\n")
```
